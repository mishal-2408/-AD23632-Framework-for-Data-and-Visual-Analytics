```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("StudentsPerformance.xls")

print(df.head())

print(df.tail())

print(df.shape)

print(df.columns.tolist())

print(df.dtypes)

print(df.info())

print(df.describe())

math_above_70 = df[df["math score"] > 70]
print(math_above_70)

female_math_above_70 = df[(df["gender"] == "female") & (df["math score"] > 70)]
print(female_math_above_70)

female_master = df[(df["gender"] == "female") & (df["test preparation course"] == "completed")]
print(female_master)

selected_columns = df[["gender", "math score"]]
print(selected_columns)

print(df.describe(include="all"))

subject_means = df[["math score", "reading score", "writing score"]].mean()

plt.figure(figsize=(8, 5))
plt.bar(subject_means.index, subject_means.values)
plt.xlabel("Subjects")
plt.ylabel("Average Score")
plt.title("Average Scores per Subject")
plt.show()

plt.figure(figsize=(8, 5))
plt.hist(df["math score"], bins=10, edgecolor="black")
plt.xlabel("Math Score")
plt.ylabel("Frequency")
plt.title("Distribution of Math Scores")
plt.show()

plt.figure(figsize=(8, 5))
plt.boxplot(df["math score"])
plt.ylabel("Math Score")
plt.title("Spread of Math Scores")
plt.show()

mean = df["math score"].mean()
median = df["math score"].median()
mode = df["math score"].mode()[0]

plt.figure(figsize=(8, 5))
plt.hist(df["math score"], bins=10, edgecolor="black")
plt.axvline(mean, linestyle="--", linewidth=2, label=f"Mean = {mean:.2f}")
plt.axvline(median, linestyle="-.", linewidth=2, label=f"Median = {median:.2f}")
plt.axvline(mode, linestyle=":", linewidth=2, label=f"Mode = {mode}")
plt.xlabel("Math Score")
plt.ylabel("Frequency")
plt.title("Math Scores with Mean, Median and Mode")
plt.legend()
plt.show()
```
