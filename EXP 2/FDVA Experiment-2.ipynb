```python
import sqlite3
import pandas as pd
```

```python
excel_file = "heart stalog dataset.xlsx"

df = pd.read_excel(excel_file)

print("Heart Statlog dataset loaded successfully.")
print()
```

```python

print("First 5 records:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())
```

```python
conn = sqlite3.connect("heart_stalog.db")

cursor = conn.cursor()

print("\nSQLite database connected successfully.")
```

```python
df.columns = [
    column.strip()
    .lower()
    .replace(" ", "_")
    .replace("-", "_")
    for column in df.columns
]
```

```python
df.to_sql(
    "heart_stalog",
    conn,
    if_exists="replace",
    index=False
)
```

```python
print("\nTable Structure:")

cursor.execute("PRAGMA table_info(heart_stalog)")

columns = cursor.fetchall()

for column in columns:
    print(column)
```

```python
print("\nFirst 10 records from SQLite:")

cursor.execute("""
SELECT *
FROM heart_stalog
LIMIT 10
""")

records = cursor.fetchall()

for record in records:
    print(record)
```

```python
cursor.execute("""
SELECT COUNT(*)
FROM heart_stalog
""")

count = cursor.fetchone()[0]

print("\nTotal number of records:", count)
```

```python
print("\nExample SQL Query:")

cursor.execute("""
SELECT *
FROM heart_stalog
LIMIT 5
""")

result = cursor.fetchall()

for row in result:
    print(row)
```

```python
conn.close()

print("\nSQLite database connection closed.")
print("Heart Statlog dataset successfully implemented in SQLite3.")
```
